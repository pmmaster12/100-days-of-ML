# Using Iterative Imputer Scikit-Learn class

```python
from sklearn.impute import IterativeImputer
from sklearn.linear_model import LinearRegression
```

```python
# -----------------------------
# Step 2: Create IterativeImputer
# -----------------------------

imputer = IterativeImputer(
    estimator=LinearRegression(),
    max_iter=10,
    random_state=42,
    initial_strategy="mean",
    imputation_order="ascending",
    verbose=3     # Shows what happens internally
)
```

```python
df_new = df.copy()
```

```python
df_new
```

```python
# -----------------------------
# Step 3: Fit + Transform
# -----------------------------

imputed_array = imputer.fit_transform(df_new)
```

```python
imputed_array
```

```python
# Convert back to DataFrame
imputed_df = pd.DataFrame(
    imputed_array,
    columns=df.columns,
    index=df.index
)
```

```python
print("\nImputed Data")
print(imputed_df)
```

```python
print("\nSequence of fitted estimators:\n")

for i, triplet in enumerate(imputer.imputation_sequence_):
    print(f"Step {i+1}:")
    print(triplet)
    print()
```
